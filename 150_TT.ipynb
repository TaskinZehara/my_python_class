## WRITE A PROGRAMM TO CALCULATE THE AREA OF A TRIANGLE WITH INPUT FROM  THE USER IN FLOAT OF THREE SIDE THE FORMULA FOR THE AREA IS GIVEN BELOW:-
- $$ S = (A + B + C)/2 $$
- $$ AREAOfTRIANGLE = \sqrt (S(S-A)(S-B)(S-C)) $$
 ```
test case :-
a=5
b= 5
c=5
 The area is 10.83
 ```

````python
import math
#a=5
#b=5c=5
a= float(input("enter side 1"))
b= float(input("enter side 2"))
c= float(input("enter side 3"))
s=(a+b+c)/2
area=math.sqrt(s*(s-a)*(s-b)*(s-c))
print(f" the area of tri with sides {a},{b},{c} is{area:.2f}")
````
