```python
import math
# Write a Python program to find the area of Triangle.
base = int(input("enter base"))
hight = int(input("enter hight"))
area=(1/2)*base*hight
print(int(area))
# heron's triangle
a = float(input("Enter first side (a): "))
b = float(input("Enter second side (b): "))
c = float(input("Enter third side (c): "))
s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s-c))
print(f"The area of the triangle is: {area:.2f}")
```

```python
# Write a Python program to calculate the area of a trapezoid.
a = float(input("Enter the length of first base (a): "))
b = float(input("Enter the length of second base (b): "))
h = float(input("Enter the perpendicular height (h): "))
area = 0.5 * (a + b) * h
print(f"The area of the trapezoid is: {area:.2f}")
```
